```python
s="communications"
dist={}
for word in s:
    if word in dist:
        dist[word]+=1
    else:
        dist[word] =1
        
print(dist)
for key,value in dist.items():
    print(f"{key}=>{value}")
```

```python
s="communications"
dist={}
for word in s:
    dist[word] = dist.get(word,0) +1
        
print(dist)
for keys,values in dist.items():
    print(f"{keys}=>{values}")
```
