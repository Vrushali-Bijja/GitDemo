```python
l=[0,2,0,4,1,3,2,0,9,0]
zeros=[]
print(len(l))
i=0
while i < len(l):
   if l[i] == 0:
       zeros.append(l[i])
       l.pop(i)
   else:
       i+=1

print(l)
print(zeros)
for zero in zeros:
    l.append(zero)
print(l)
```

```python
l=[0,2,0,4,1,3,2,0,9,0]
unique = []
for num in l:
    if num not in unique:
        unique.append(num)
print(unique)
sorted_list = sorted(unique)
print(sorted_list[len(unique)-2])
```
