```python
s = "This is not a Test Test is not This"
s.lower()
words=s.split(" ")
word_count={}

for word in words:
    word_count[word] = word_count.get(word,0)+1
for key, value in word_count.items():
    print(key, ":", value)
```

```python
string= "Mohit"
words={}
for ch in string.lower():
    print(ch)
    words[ch] = words.get(ch,0)+1
print(words)

for keys,values in words.items():
    print(f"{keys}->{values}")
```
